### Make Program more Dynamicn using SQlite

```python
    import sqlite3
    from datetime import date
    from pathlib import Path
    
    # =================================================
    # DATABASE CONFIGURATION
    # =================================================
    
    DATA_DIR = Path("data")
    DATA_DIR.mkdir(exist_ok=True)
    
    DB_PATH = DATA_DIR / "expenses.db"
    
    # =================================================
    # DATABASE CONNECTION
    # =================================================
    
    def get_connection():
        return sqlite3.connect(DB_PATH)
    
    # =================================================
    # CREATE DATABASE TABLE
    # =================================================
    
    def create_table():
        connection = get_connection()
    
        cursor = connection.cursor()
    
        cursor.execute("""
            CREATE TABLE IF NOT EXISTS expenses (
                id INTEGER PRIMARY KEY AUTOINCREMENT,
                date TEXT NOT NULL,
                category TEXT NOT NULL,
                description TEXT NOT NULL,
                amount REAL NOT NULL
            )
        """)
    
        connection.commit()
        connection.close()
    
    
    # ====================================================
    # ADD EXPENSE
    # ====================================================
        
    def add_expense():
        category = input("Enter category: ").strip()
        description = input("Enter description: ").strip()
    
        while True:
    
            try:
                amount = float(
                    input("Enter amount: ")
                )
    
                if amount <= 0:
                    print("Amount must be greater than zero.")
                    continue
                break
    
            except ValueError:
                print("Please enter a valid number.")
    
        connection = get_connection()
        cursor = connection.cursor()
    
        cursor.execute("""
            INSERT INTO expenses
            (date, category, description, amount)
            VALUES (?, ?, ?, ?)
        """, (
            str(date.today()),
            category,
            description,
            amount
        ))
    
        connection.commit()
        connection.close()
    
        print("Expense added successfullly.")
        print("Expense saved to SQLite data. ")

    # ====================================================
    # VIEW EXPENSE
    # ====================================================

    def view_expenses():

        connection = get_connection()
        cursor = connection.cursor()

        cursor.execute("""
            SELECT
                id,
                date,
                category,
                description,
                amount
            FROM expenses
            ORDER BY id
        """)

        expenses = cursor.fetchall()

        connection.close()

        if not expenses:
            print("No expenses found.")
            return

        print("\nID |  Date       | Category         | Description            | Amount")
        print("-" * 75)

        for expense in expenses:

            expense_id = expense[0]
            expense_date = expense[1]
            category = expense[2]
            description = expense[3]
            amount = expense[4]

            print(
                f"{expense_id}  | "
                f"{expense_date}  | "
                f"{category}             | "
                f"{description}                  | "
                f"${amount:.2f}"

            )


    # ============================================================
    # UPDATE EXPENSE
    # ============================================================
    
    def update_expense():
    
        try:
            expense_id = int(
                input("Enter expense ID to update: ")
            )
    
        except ValueError:
            print("Invalid ID.")
            return
    
    
        connection = get_connection()
    
        cursor = connection.cursor()
    
    
        # --------------------------------------------------------
        # First retrieve the existing expense
        # --------------------------------------------------------
    
        cursor.execute("""
            SELECT
                id,
                date,
                category,
                description,
                amount
            FROM expenses
            WHERE id = ?
        """, (expense_id,))
    
    
        expense = cursor.fetchone()
    
    
        if expense is None:
    
            print("Expense not found.")
    
            connection.close()
    
            return
    
    
        # --------------------------------------------------------
        # Display current values
        # --------------------------------------------------------
    
        print("\nCurrent Expense")
        print("----------------")
        print(f"Category: {expense[2]}")
        print(f"Description: {expense[3]}")
        print(f"Amount: ${expense[4]:.2f}")
    
        print(
            "\nLeave a field empty "
            "to keep the current value."
        )
    
    
        # --------------------------------------------------------
        # Get new values
        # --------------------------------------------------------
    
        category = input(
            f"Category [{expense[2]}]: "
        ).strip()
    
        description = input(
            f"Description [{expense[3]}]: "
        ).strip()
    
        amount_input = input(
            f"Amount [{expense[4]}]: "
        ).strip()
    
    
        # --------------------------------------------------------
        # Keep existing values if user leaves blank
        # --------------------------------------------------------
    
        if not category:
            category = expense[2]
    
        if not description:
            description = expense[3]
    
    
        if amount_input:
    
            try:
    
                amount = float(amount_input)
    
                if amount <= 0:
                    print(
                        "Amount must be greater than zero."
                    )
    
                    connection.close()
    
                    return
    
            except ValueError:
    
                print("Invalid amount.")
    
                connection.close()
    
                return
    
        else:
    
            amount = expense[4]
    
    
        # --------------------------------------------------------
        # UPDATE DATABASE
        # --------------------------------------------------------
    
        cursor.execute("""
            UPDATE expenses
            SET
                category = ?,
                description = ?,
                amount = ?
            WHERE id = ?
        """, (
            category,
            description,
            amount,
            expense_id
        ))
    
    
        connection.commit()
    
        connection.close()
    
    
        print("Expense updated successfully.")
        print("Changes saved to SQLite database.")


    # ====================================================
    # DELETE PROGRAM
    # ====================================================

    def delete_expense():

        try:

            expense_id = int(
                input("Enter expense ID to delete: ")
            )

        except ValueError:

            print("Invalid ID.")

            return

        connection  = get_connection()
        cursor = connection.cursor()

        #------------------------------------------------
        # check if expense exists
        # -----------------------------------------------

        cursor.execute("""
            SELECT *
            FROM expenses
            WHERE id = ?
        """, (expense_id, 1))

        expense = cursor.fetchone()

        if expense is None:
            print("Expense not found.")

            connection.close()

            return
        # Ask for Confimation

        print("\nExpense to delete:")
        print(f"ID: {expense[0]}")
        print(f"Date: {expense[1]}")
        print(f"Category: {expense[2]}")
        print(f"Description: {expense[3]}")
        print(f"Amount: ${expense[4]:.2f}")


        confirmation = input(
            "Are you sure you want to delete it? (y/n): "
        ).strip().lower()


        if confirmation != "y":
            print("Delete cancelled.")

            connection.close()

            return

    # Delete From Database

        cursor.execute("""
            DELETE FROM expenses
            WHERE id = ?
        """, (expense_id,))
    
        connection.commit()
    
        connection.close()
    
        print("Expense deleted successfully.")
        print("Changes saved to SQLite database.")


    # ====================================================
    # CALCULATE TOTAL
    # ====================================================

    def calculate_total():

        connection = get_connection()

        cursor = connection.cursor()

        cursor.execute("""
            SELECT SUM(amount)
            FROM expenses
        """)

        result = cursor.fetchone()

        connection.close()

        total = result[0]

        if total is None:
            total = 0

        print(
            f"Total expenses: ${total:.2f}"
        )

    
    # ====================================================
    # CALCULATE CATEGORY TOTAL
    # ====================================================

    def calculate_category_total():

        category = input(
            "Enter category: "
        ).strip()

        connection = get_connection()
        cursor = connection.cursor()

        cursor.execute("""
            SELECT SUM(amount)
            FROM expenses
            WHERE LOWER(category) = LOWER(?)
        """, (category,))


        result = cursor.fetchone()

        connection.close()

        total = result[0]

        if total is None:
            total =0

        print(
            f"Total for {category}: ${total:.2f}"
        )

    # ====================================================
    # CALCULATE MONTHLY TOTAL
    # ====================================================

    def calculate_monthly_total():

        month = input(
            "Enter month (YYYY-MM): "
        ).strip()

        connection = get_connection()
        cursor = connection.cursor()

        cursor.execute("""
            SELECT SUM(amount)
            FROM expenses
            WHERE date LIKE ?
        """, (f"{month}%",))

        result = cursor.fetchone()

        connection.close()

        total = result[0]

        if total is None:
            total =0

        print(
            f"Total for {month}: ${total:2f}"
        )
    
    # ====================================================
    # MAIN PROGRAM
    # ====================================================

    def main():

        # Create database/table if they do not exit
        create_table()

        while True:

            print("\n=============================")
            print("      PYTHON EXPENSE TRACKER")
            print("===============================")

            print("1. Add expense")
            print("2. View expenses")
            print("3. Update expense")
            print("4. Delete expense")
            print("5. Calculate total")
            print("6. Calculate category total")
            print("7. Calculate monthly total")
            print("8. Exit")

            choice = input(
                "\nChoose an option: "
            ).strip()

            if choice == "1":
                add_expense()

            elif choice == "2":
                view_expenses()

            elif choice == "3":
                update_expense()

            elif choice == "4":
                delete_expense()

            elif choice == "5":
                calculate_total()

            elif choice == "6":
                calculate_category_total()

            elif choice == "7":
                calculate_monthly_total()

            elif choice == "8":
                print("Goodbye!")
                break

            else:
                print(
                    "Invalid option. Please choose"
                    "1 to 8."
                )
    


    # ====================================================
    # PROGRAM ENTRY POINT
    # ====================================================

    if __name__ == "__main__":
        main()
```
